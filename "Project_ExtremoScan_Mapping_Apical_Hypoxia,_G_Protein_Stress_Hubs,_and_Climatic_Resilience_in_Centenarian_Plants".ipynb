```python
# Cell 1: Environment Setup & Library Imports
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from Bio import SeqIO
from scipy.integrate import solve_ivp
from scipy.ndimage import gaussian_filter

# Configure visualization defaults
sns.set_theme(style="whitegrid")
plt.rcParams['figure.dpi'] = 300
print("ExtremoScan Environment Initialized Successfully!")
```

```python
# Cell 4: Pillar II Epigenetic Clock & Cambial Resilience Profiler
class ExtremoScanPillarIIProfiler:
    def __init__(self, age_cohorts):
        self.age_cohorts = age_cohorts

    def simulate_dal1_hypomethylation(self):
        """Simulates CHG DNA methylation percentage decay in the DAL1 intron across tree ages."""
        np.random.seed(42)
        methylation_level = 80.0 * np.exp(-0.0035 * self.age_cohorts) + 15.0 + np.random.normal(0, 1.2, len(self.age_cohorts))
        return np.clip(methylation_level, 15.0, 95.0)

    def simulate_cambial_transcriptome_shifts(self):
        """Simulates TPM expression profiles for Defense/R-genes vs Cell Division genes across tree ages."""
        np.random.seed(42)
        r_genes_tpm = 40.0 + 0.1 * self.age_cohorts + np.random.normal(0, 2.5, len(self.age_cohorts))
        differentiation_tpm = 130.0 - 0.14 * self.age_cohorts + np.random.normal(0, 3.0, len(self.age_cohorts))

        df = pd.DataFrame({
            "Age_Years": self.age_cohorts,
            "R_Genes_Defense_TPM": r_genes_tpm,
            "Cell_Differentiation_TPM": differentiation_tpm
        })
        return df

    def plot_epigenetic_and_expression_trends(self):
        ages = self.age_cohorts
        meth = self.simulate_dal1_hypomethylation()
        df_expr = self.simulate_cambial_transcriptome_shifts()

        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

        # Panel 1: DAL1 Clock
        ax1.plot(ages, meth, color='purple', lw=2.5, label='DAL1 Intron CHG Methylation (%)')
        ax1.set_title("ExtremoScan: DAL1 Epigenetic Age Timer")
        ax1.set_xlabel("Tree Age (Years)")
        ax1.set_ylabel("CHG Methylation Level (%)")
        ax1.grid(True, linestyle='--', alpha=0.5)
        ax1.legend()

        # Panel 2: Expression Shifts
        ax2.plot(ages, df_expr["R_Genes_Defense_TPM"], color='forestgreen', lw=2.5, label='R-Genes / Defense Pathway')
        ax2.plot(ages, df_expr["Cell_Differentiation_TPM"], color='darkorange', lw=2.5, label='Cell Division / Differentiation')
        ax2.set_title("Ginkgo Cambial Transcriptome Shift (Ages 15 - 650+)")
        ax2.set_xlabel("Tree Age (Years)")
        ax2.set_ylabel("Normalized Expression (TPM)")
        ax2.grid(True, linestyle='--', alpha=0.5)
        ax2.legend()

        plt.tight_layout()
        plt.savefig("extremoscan_pillar_ii_epigenetic_profiler.png", dpi=300)
        plt.show()

# Execution Example
ages_array = np.linspace(15, 650, 100)
profiler = ExtremoScanPillarIIProfiler(ages_array)
profiler.plot_epigenetic_and_expression_trends()
print("Pillar II Profiler executed successfully.")
```

```python
# Cell 6: Pillar I 2D Spatial Hypoxia PDE Mesh & ODE Kinetics Simulation
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.ndimage import gaussian_filter

def run_pillar_i_simulations():
    # --- Part 1: 2D Spatial Hypoxia PDE Mesh Simulation ---
    grid_size = 150
    o2_map = np.ones((grid_size, grid_size)) * 21.3
    center_x, center_y = grid_size // 2, grid_size // 2
    y, x = np.ogrid[:grid_size, :grid_size]
    dome_mask = (x - center_x)**2 / ((grid_size*0.35)**2) + (y - center_y)**2 / ((grid_size*0.25)**2) <= 1

    for _ in range(100):
        core_mask = dome_mask & ((x - center_x)**2 + (y - center_y)**2 < (grid_size * 0.1)**2)
        o2_map[core_mask] -= 0.15
        l2_mask = dome_mask & ~core_mask & ((x - center_x)**2 + (y - center_y)**2 < (grid_size * 0.25)**2)
        o2_map[l2_mask] -= 0.08
        o2_map[~dome_mask] = 21.3
        o2_map = gaussian_filter(o2_map, sigma=1.2)
        o2_map = np.clip(o2_map, 2.5, 21.3)

    # Plotting 2D Spatial Hypoxia Mesh
    plt.figure(figsize=(8, 6))
    plt.imshow(o2_map, cmap='viridis', origin='lower')
    plt.colorbar(label='Oxygen Partial Pressure (kPa)')
    plt.title("ExtremoScan: 2D SAM Spatial Hypoxia Mesh (~3.5 kPa Core)")
    plt.savefig("extremoscan_2d_sam_hypoxia_mesh.png", dpi=300)
    plt.show()

    # --- Part 2: CLAVATA-CT2-WUSCHEL ODE Feedback Kinetics ---
    t_span = (0, 50)
    t_eval = np.linspace(0, 50, 500)
    initial_state = [1.0, 1.0, 0.5] # [WUS, CLV3, CT2]

    def temperature_profile(t):
        return 42.0 if 10 <= t <= 30 else 25.0

    def stem_cell_feedback_system(t, y, temp_spike):
        wus, clv3, ct2 = y
        alpha_1, k_i, n, gamma_1 = 2.0, 1.5, 3, 0.5
        alpha_2, gamma_2 = 1.2, 0.8
        stress_beta = 0.05 if temp_spike > 40.0 else 0.0
        ct2_protection = 1.0 + (0.5 * ct2)

        dwus_dt = (alpha_1 / (1.0 + (clv3 / k_i)**n)) - (gamma_1 * wus) - ((stress_beta / ct2_protection) * wus)
        dclv3_dt = (alpha_2 * wus) - (gamma_2 * clv3)
        dct2_dt = 0.1 * (temp_spike - 25.0) - 0.2 * ct2 if temp_spike > 35.0 else -0.1 * ct2
        return [dwus_dt, dclv3_dt, dct2_dt]

    solution_wus, solution_clv3, solution_ct2 = [], [], []
    dt = t_eval[1] - t_eval[0]
    current_y = initial_state

    for t in t_eval:
        temp = temperature_profile(t)
        res = solve_ivp(lambda tp, yp: stem_cell_feedback_system(tp, yp, temp), [0, dt], current_y, t_eval=[dt])
        current_y = res.y[:, -1]
        solution_wus.append(current_y[0])
        solution_clv3.append(current_y[1])
        solution_ct2.append(current_y[2])

    # Plotting ODE Resilience Dynamics
    plt.figure(figsize=(10, 5))
    plt.plot(t_eval, solution_wus, label='WUSCHEL (Stem Cell Maintenance)', color='green', lw=2)
    plt.plot(t_eval, solution_clv3, label='CLAVATA3 (Feedback Inhibitor)', color='blue', lw=2)
    plt.plot(t_eval, solution_ct2, label='CT2 G-alpha (Stress Buffer)', color='orange', lw=2)
    plt.axvspan(10, 30, color='red', alpha=0.15, label='Super El Niño Thermal Spike (>40°C)')
    plt.title("ExtremoScan: CLAVATA-CT2-WUSCHEL Dynamic ODE Feedback under Thermal Stress")
    plt.xlabel("Time Units")
    plt.ylabel("Normalized Concentration / Activity")
    plt.legend()
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.tight_layout()
    plt.savefig("extremoscan_ode_feedback_kinetics.png", dpi=300)
    plt.show()
    print("Pillar I Spatial Mesh and ODE Simulations executed successfully!")

# Run Pillar I simulations
run_pillar_i_simulations()
```
